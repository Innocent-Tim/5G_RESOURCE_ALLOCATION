```python
import torch

# Create a tensor from a list
a = torch.tensor([10, 20, 30])

# Create a tensor filled with zeros
b = torch.zeros(3)

# Create a tensor filled with ones
c = torch.ones(3)

# Create a 2-by-3 tensor of random values
d = torch.rand(2, 3)

print("Tensor a:", a)
print("Zeros:", b)
print("Ones:", c)
print("Random tensor:\n", d)
```

```python
import torch

a = torch.tensor([2.0, 4.0, 6.0])
b = torch.tensor([1.0, 3.0, 5.0])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Multiply by 2:", a * 2)
```

###Welcome to Lesson 4! You're making good progress. So far, you've learned how to create tensors, select elements, and perform calculations. Now we're going to learn one of the most important concepts in deep learning: how PyTorch calculates gradients so a model can learn from its mistakes.
1. What is a gradient?
Imagine you're training a machine learning model to predict internet bandwidth requirements. Its predictions are initially inaccurate, so it needs to adjust its internal parameters to reduce its errors.

A gradient tells us how a quantity changes when an input or parameter changes. During training, gradients help the model determine how to adjust its parameters to reduce the loss.

- The height of the hill represents the loss.

- Gradient descent uses gradients to adjust parameters toward lower loss.

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

print("x:", x)
print("y:", y)
```
